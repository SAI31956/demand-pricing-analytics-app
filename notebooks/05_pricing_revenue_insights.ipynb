```python
import numpy as np
import pandas as pd
def revenue(price, demand):
    return price * demand
```

```python
import numpy as np
import pandas as pd

def find_optimal_price(
    base_features,
    model,
    cost,
    price_range
):
    results = []

    for price in price_range:
        features = base_features.copy()
        features['price'] = price

        predicted_demand = model.predict(
            pd.DataFrame([features])
        )[0]

        predicted_demand = max(predicted_demand, 0)

        profit = (price - cost) * predicted_demand

        results.append((price, predicted_demand, profit))

    results_df = pd.DataFrame(
        results,
        columns=['price', 'predicted_demand', 'profit']
    )

    return results_df.sort_values(by='profit', ascending=False)
```

```python
def dynamic_price(base_price, inventory, competitor_price):
    if inventory < 50:
        return base_price * 1.08
    elif competitor_price < base_price:
        return competitor_price - 1
    return base_price
# A simple dynamic pricing function based on inventory and competitor pricing
```

```python
import pandas as pd
df = pd.read_csv('../data/processed_data/ecommerce_pricing_clean.csv')
df['simulated_price'] = df.apply(
    lambda x: dynamic_price(
        x['price'], x['inventory_level'], x['competitor_price']
    ),
    axis=1
)

df[['price', 'simulated_price']].head()
```
