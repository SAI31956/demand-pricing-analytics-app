```python
import pandas as pd

df = pd.read_csv('../data/raw_data/ecommerce_pricing_raw_data.csv')
df.head()
```

```python
df.isnull().sum()
df.duplicated().sum()
```

```python
df['profit'] = df['revenue'] - (df['cost'] * df['units_sold'])
df['margin_pct'] = df['profit'] / df['revenue']
df['price_diff_competitor'] = df['price'] - df['competitor_price']
```

```python
df.to_csv('../data/processed_data/ecommerce_pricing_clean.csv', index=False)
```
